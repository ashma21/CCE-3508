```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
import json
import time

# ==========================================
# 1. SYNTHETIC DATA GENERATION (CLIUS MODEL)
# ==========================================
# Simulates 30 days of hourly energy data (720 hours)
np.random.seed(42)
hours = 720
time_stamps = pd.date_range(start="2026-01-01", periods=hours, freq="h")

# Base load + cyclic shift patterns (kW) + Gaussian noise
energy_kw = 100 + 50 * np.sin(2 * np.pi * time_stamps.hour / 24) + np.random.normal(0, 5, hours)

# Waste Coefficient (lambda) = 0.15 kg/kWh + noise
lambda_coeff = 0.15
noise = np.random.normal(0, 2, hours)

# Mathematical Model Equation: W_predicted = E_c * lambda + noise
waste_kg = (energy_kw * lambda_coeff) + noise

df = pd.DataFrame({"Timestamp": time_stamps, "Energy_kW": energy_kw, "Waste_kg": waste_kg})
print("--- Data Sample ---")
print(df.head())

# ==========================================
# 2. DATA PREPARATION FOR LSTM MODEL
# ==========================================
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df[["Energy_kW", "Waste_kg"]])

def create_sequences(data, seq_length=24):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length, 0])  # Input: Energy history (past 24 hrs)
        y.append(data[i+seq_length, 1])     # Target: Next hour waste volume
    return np.array(X), np.array(y)

SEQ_LEN = 24
X, y = create_sequences(scaled_data, SEQ_LEN)
X = np.reshape(X, (X.shape[0], X.shape[1], 1))

# 80-20 Train-Test Split
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# ==========================================
# 3. EDGE-AI LSTM MODEL DEVELOPMENT
# ==========================================
model = Sequential([
    LSTM(32, return_sequences=False, input_shape=(SEQ_LEN, 1)),
    Dropout(0.1),
    Dense(16, activation='relu'),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
print("\n--- Training Lightweight Edge-AI LSTM ---")
history = model.fit(X_train, y_train, epochs=15, batch_size=16, validation_split=0.1, verbose=1)

# ==========================================
# 4. EVALUATION METRICS
# ==========================================
y_pred_scaled = model.predict(X_test)
# Inverse scale predictions for actual kg evaluation
dummy_pred = np.zeros((len(y_pred_scaled), 2))
dummy_pred[:, 1] = y_pred_scaled.flatten()
y_pred_actual = scaler.inverse_transform(dummy_pred)[:, 1]

dummy_true = np.zeros((len(y_test), 2))
dummy_true[:, 1] = y_test
y_true_actual = scaler.inverse_transform(dummy_true)[:, 1]

rmse = np.sqrt(mean_squared_error(y_true_actual, y_pred_actual))
mape = mean_absolute_percentage_error(y_true_actual, y_pred_actual) * 100

print("\n--- Evaluation Results ---")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f} kg")
print(f"Mean Absolute Percentage Error (MAPE): {mape:.2f}%")

# ==========================================
# 5. TFLITE QUANTIZATION (FOR ESP32-S3 EDGE DEPLOYMENT)
# ==========================================
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_quant_model = converter.convert()

# Save edge model binary file
with open("clius_edge_model.tflite", "wb") as f:
    f.write(tflite_quant_model)
print(f"\nQuantized TFLite Model Saved. File Size: {len(tflite_quant_model) / 1024:.2f} KB")

# ==========================================
# 6. G2G MQTT TOKEN GENERATOR
# ==========================================
def generate_waste_readiness_token(factory_id, predicted_waste, surge_threshold=20.0):
    is_surge = predicted_waste > surge_threshold
    token = {
        "factory_id": factory_id,
        "timestamp": int(time.time()),
        "predicted_waste_kg": round(float(predicted_waste), 2),
        "surge_warning": bool(is_surge),
        "dispatch_window_hrs": 2 if is_surge else 6
    }
    return json.dumps(token)

sample_prediction = y_pred_actual[-1]
token_payload = generate_waste_readiness_token("FAC_IIUC_01", sample_prediction)
print("\n--- Generated MQTT Waste-Readiness Token ---")
print(token_payload)
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
import os
import json
import time
import matplotlib.pyplot as plt

# ==========================================
# 1. SYNTHETIC DATA GENERATION (CLIUS MODEL)
# ==========================================
np.random.seed(42)
tf.random.set_seed(42)

hours = 720
time_stamps = pd.date_range(start="2026-01-01", periods=hours, freq="h")

# Energy load + waste calculation model
energy_kw = 100 + 50 * np.sin(2 * np.pi * time_stamps.hour / 24) + np.random.normal(0, 5, hours)
lambda_coeff = 0.15
noise = np.random.normal(0, 2, hours)
waste_kg = (energy_kw * lambda_coeff) + noise

df = pd.DataFrame({"Timestamp": time_stamps, "Energy_kW": energy_kw, "Waste_kg": waste_kg})
print("--- Data Sample ---")
print(df.head())

# ==========================================
# 2. DATA PREPARATION FOR LSTM MODEL
# ==========================================
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df[["Energy_kW", "Waste_kg"]])

def create_sequences(data, seq_length=24):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length, 0])  # Energy input history
        y.append(data[i+seq_length, 1])     # Next hour waste target
    return np.array(X), np.array(y)

SEQ_LEN = 24
X, y = create_sequences(scaled_data, SEQ_LEN)
X = np.reshape(X, (X.shape[0], X.shape[1], 1))

split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# ==========================================
# 3. EDGE-AI LSTM MODEL DEVELOPMENT
# ==========================================
# Note: Input layer explicitly defined and unroll=True set to fix TFLite conversion
model = Sequential([
    Input(shape=(SEQ_LEN, 1)),
    LSTM(32, return_sequences=False, unroll=True),
    Dropout(0.1),
    Dense(16, activation='relu'),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
print("\n--- Training Lightweight Edge-AI LSTM ---")
history = model.fit(X_train, y_train, epochs=15, batch_size=16, validation_split=0.1, verbose=1)

# ==========================================
# 4. EVALUATION METRICS
# ==========================================
y_pred_scaled = model.predict(X_test)

dummy_pred = np.zeros((len(y_pred_scaled), 2))
dummy_pred[:, 1] = y_pred_scaled.flatten()
y_pred_actual = scaler.inverse_transform(dummy_pred)[:, 1]

dummy_true = np.zeros((len(y_test), 2))
dummy_true[:, 1] = y_test
y_true_actual = scaler.inverse_transform(dummy_true)[:, 1]

rmse = np.sqrt(mean_squared_error(y_true_actual, y_pred_actual))
mape = mean_absolute_percentage_error(y_true_actual, y_pred_actual) * 100

print("\n==========================================")
print("--- EVALUATION RESULTS FOR YOUR PAPER ---")
print("==========================================")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f} kg")
print(f"Mean Absolute Percentage Error (MAPE): {mape:.2f}%")

# ==========================================
# 5. TFLITE QUANTIZATION & MODEL SAVING
# ==========================================
# Save full Keras model
model.save("clius_lstm_model.keras")

# Convert to TFLite with custom op handling fallback
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.target_spec.supported_ops = [
    tf.lite.OpsSet.TFLITE_BUILTINS,
    tf.lite.OpsSet.SELECT_TF_OPS
]
converter._experimental_lower_tensor_list_ops = False

tflite_quant_model = converter.convert()

with open("clius_edge_model.tflite", "wb") as f:
    f.write(tflite_quant_model)

size_kb = os.path.getsize("clius_edge_model.tflite") / 1024
print(f"Quantized TFLite Model Saved: 'clius_edge_model.tflite'")
print(f"Model Memory Footprint: {size_kb:.2f} KB")

# ==========================================
# 6. GENERATE PLOT FOR LATEX PAPER
# ==========================================
plt.figure(figsize=(9, 4))
plt.plot(y_true_actual[:72], label="Actual Industrial Waste (kg)", color="#1f77b4", linewidth=2)
plt.plot(y_pred_actual[:72], label="Edge-AI Predicted Waste (kg)", color="#ff7f0e", linestyle="--", linewidth=2)
plt.title("CLIUS Framework: 72-Hour Predictive Waste Tracking")
plt.xlabel("Time (Hours)")
plt.ylabel("Waste Output (kg)")
plt.legend(loc="upper right")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.savefig("clius_prediction_graph.png", dpi=300)
plt.show()

print("\n✅ Download 'clius_prediction_graph.png' from Colab left folder icon to upload to Overleaf!")
```
