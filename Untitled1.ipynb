```python
import numpy as np
import tensorflow as tf

# 1. Generate Synthetic Sequential Energy & Waste Data
np.random.seed(42)
num_samples = 1000
time_steps = 10
features = 2

X_train = np.random.rand(num_samples, time_steps, features).astype(np.float32)
y_train = (X_train[:, -1, 0] * 12.5 + X_train[:, -1, 1] * 3.2 + np.random.normal(0, 0.5, num_samples)).astype(np.float32)

# 2. Build Model with UNROLLED LSTM and STATIC INPUT SHAPE
model = tf.keras.Sequential([
    tf.keras.layers.InputLayer(batch_input_shape=(1, time_steps, features)),
    tf.keras.layers.LSTM(16, return_sequences=False, unroll=True),  # unroll=True fixes TensorList error
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y_train, epochs=5, batch_size=1, verbose=0)

# 3. Representative Dataset Generator for INT8 Quantization
def representative_data_gen():
    for i in range(100):
        # Must yield batch size of 1 matching batch_input_shape
        yield [X_train[i:i+1]]

# 4. Convert and Quantize Model to TensorFlow Lite (INT8)
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_data_gen
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.int8
converter.inference_output_type = tf.int8

tflite_quant_model = converter.convert()

# 5. Save the Quantized Binary File
with open("clius_model_quant.tflite", "wb") as f:
    f.write(tflite_quant_model)

print(f"Success! Model size: {len(tflite_quant_model) / 1024:.2f} KB")

# 6. Automatically Generate C Header File (clius_model_data.h)
hex_bytes = [f"0x{b:02x}" for b in tflite_quant_model]
array_body = ", ".join(hex_bytes)

c_header_content = f"""#ifndef CLIUS_MODEL_DATA_H
#define CLIUS_MODEL_DATA_H

const unsigned char clius_model_quant_tflite[] = {{
    {array_body}
}};
const unsigned int clius_model_quant_tflite_len = {len(tflite_quant_model)};

#endif // CLIUS_MODEL_DATA_H
"""

with open("clius_model_data.h", "w") as f:
    f.write(c_header_content)

print("Generated 'clius_model_data.h' for ESP32-S3 deployment!")
```
